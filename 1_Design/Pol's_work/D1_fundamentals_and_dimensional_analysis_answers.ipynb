## Some simple questions

What is the main difference between a **driven turbomachine** (Power-absorving) and a **driving turbomachine** (Power-Generating)? Give some examples of both types. How is that related to the **Euler Equation**?

The **driven turbomachine** are those that _absorb_ power to increase the fluid pressure or head. For this work absorbing machines the work rate must be less than 0. The **driving turbomachine** are those that _produce_ power by expanding fluid to a lower pressure or head. For this work producing machines the work rate must be more than 0. The driven turbomachine are ducted and unducted fans, compressors, and pumps. Driving turbomachine are wind, hydraulic, steam, and gas turbines. This distinction follows directly from Euler’s turbomachinery equation. Using the convention that positive work is work done by the shaft on the fluid. For compressors and pumps, 𝑤>0: power is transferred from the shaft to the fluid. For turbines, w<0: power is transferred from the fluid to the shaft. If turbine work is defined as the positive power extracted from the fluid, the equation must be written reversed.

Explain with your words what is the _rothalpy_. Why is that conserved in a rotor?



The rothalpy is a variation of the stagnation enthalpy, but implementating the rotating movement of the turbomachine. When the rotor gives energy to the fluid, the fluid’s stagnation enthalpy increases, but the angular momentum term increases by exactly the same amount. Their difference therefore remains constant. For a turbine, the fluid loses stagnation enthalpy and gives work to the rotor, but the same balance still applies.

```python
# TODO: Compute velocities and angles, and draw the triangles of velocities. Apply Euler equation to estimate head pressure. Compute efficiency.
D_h = 0.064 #m
D_t = 0.250
r_h = D_h / 2
r_t = D_t / 2
r_rms = (0.5*(r_h**2 + r_t**2))**0.5
print(f"r_rms = {r_rms : 4f}")
```

For the previous fan, compute: 
- flow, head and power coefficients
$$ \phi = \frac{Q}{\omega D^3}$$
$$ \psi = \frac{\Delta p_0}{\rho \omega^2 D^2}$$
$$ \hat{P} = \frac{P_h}{\rho \omega^3 D^5} = \phi \psi$$
- Reynolds number
$$ \text{Re}_D = \frac{\rho c_m D}{\mu} $$
- Specific velocity and diameter
$$ D_s = \frac{D(\frac{\Delta p_0}{\rho})^{1/4}}{Q^{1/2}} $$
$$ \Omega_s = \frac{\Omega Q^{1/2}}{(\frac{\Delta p_0}{\rho})^{3/4}} $$
- Locate the fan into the Cordier diagram 

```python
# TODO: Compute dimnesionless parameters. Locate in the Cordier diagram
import numpy as np

Dt = 0.250
Dh = 0.064
rt = Dt/2
rh = Dh/2

rm = np.sqrt(0.5*(rh**2+rt**2))
omega = 2275 * np.pi / 30
Um = omega * rm
A = np.pi * (rt**2 - rh**2)
Q = 1617 / 3600
cx = Q/A
betal = np.degrees(np.arctan(Um/cx))
print(rm, Um, cx, betal)
```
